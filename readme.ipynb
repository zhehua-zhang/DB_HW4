## Usage

All commands run with `run_etl.py`

first set envriomental variable

`$env:HW4_MYSQL_DSN="mysql+pymysql://root:@127.0.0.1:3306/sakila"`
 
this is just my database, run with root and no password

### Full-load 

load all source data

`python run_etll.py --src-dsn "%HW4_MYSQL_DSN%" --tgt-dsn "sqlite:///analytics.sqlite3" full-load`

### Incremental

load only new or changed data

`python run_etll.py --src-dsn "%HW4_MYSQL_DSN%" --tgt-dsn "sqlite:///analytics.sqlite3" incremental`

`python run_etll.py --src-dsn "%HW4_MYSQL_DSN%" --tgt-dsn "sqlite:///analytics.sqlite3" incremental --batch 2000 --backfill-days 30`


### Validate

`python run_etll.py --src-dsn "%HW4_MYSQL_DSN%" --tgt-dsn "sqlite:///analytics.sqlite3" validate --days 30`
